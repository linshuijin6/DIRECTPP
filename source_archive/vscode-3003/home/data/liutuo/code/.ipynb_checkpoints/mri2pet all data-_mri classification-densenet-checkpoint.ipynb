```python
import os
import pandas as pd
import torch
from transformers import AutoProcessor, AutoModel
import scipy.io as sio

# 读取CSV文件并创建Subject到Description和Group的映射
label_path1 = "/home/liushiyu/liutuo-test/ADNI1+2_baseline/ADNI1_imdb_18m_with_descriptions.csv"
label_path2 = "/home/liushiyu/liutuo-test/ADNI1+2_baseline/ADNI2_imdb_18m_with_descriptions.csv"

label1 = pd.read_csv(label_path1)
label2 = pd.read_csv(label_path2)

label1_dict = dict(zip(label1["Subject"], label1["Group"]))
label2_dict = dict(zip(label2["Subject"], label2["Group"]))

label1_desc_dict = dict(zip(label1["Subject"], label1["Description"]))
label2_desc_dict = dict(zip(label2["Subject"], label2["Description"]))

# 使用本地路径加载处理器和模型
local_model_path = "/home/ssd/liutuo/BiomedCLIP"
processor = AutoProcessor.from_pretrained(local_model_path, trust_remote_code=True)
model = AutoModel.from_pretrained(local_model_path, trust_remote_code=True)

device = torch.device('cuda') if torch.cuda.is_available() else torch.device('cpu')
model.to(device)
model.eval()

# 定义函数以读取.mat文件
def mat_load(mat_file_dir):
    mat = sio.loadmat(mat_file_dir)
    image = mat["IMG"]
    return image

# 获取路径中的数据列表
def get_data_list_from_path(path):
    data_list = [os.path.join(path, data) for data in os.listdir(path)]
    return data_list

# 获取路径中的数据并创建映射
def get_data_from_path(path, label_dict):
    data_all = []
    for data in os.listdir(path):
        image = os.path.join(path, data)
        name = data[:-4]
        try:
            label = label_dict[name]
        except KeyError:
            continue
        data_all.append({"image": image, "name": name, "label": label})
    return data_all

# 获取配对数据并附加Description
def get_data_from_path_paired_with_description(path, label_dict, label_desc_dict, mri_name_image_dict):
    data_all = []
    for data in os.listdir(path):
        pet = os.path.join(path, data)
        name = data[:-4]
        try:
            mri = mri_name_image_dict[name]
            label = label_dict[name]
            description = label_desc_dict[name]
        except KeyError:
            continue
        data_all.append({
            "pet": pet,
            "mri": mri,
            "name": name,
            "label": label,
            "description": description,
        })
    return data_all

# 设置路径
mri_path = "/home/liushiyu/liutuo-test/ADNI1+2_baseline/MRI"
pet_path = "/home/liushiyu/liutuo-test/ADNI1+2_baseline/PET"

# 获取MRI数据并创建映射
data_all = get_data_from_path(mri_path, label1_dict)
data_all_adni2 = get_data_from_path(mri_path, label2_dict)

mri_name_image_dict = dict(zip([data["name"] for data in data_all],
                               [data["image"] for data in data_all]))
mri_name_image_dict_adni2 = dict(zip([data["name"] for data in data_all_adni2],
                                     [data["image"] for data in data_all_adni2]))

# 获取配对数据并附加Description
paired_all_with_desc1 = get_data_from_path_paired_with_description(pet_path, label1_dict, label1_desc_dict, mri_name_image_dict)
paired_all_with_desc2 = get_data_from_path_paired_with_description(pet_path, label2_dict, label2_desc_dict, mri_name_image_dict_adni2)

# 合并两个数据集
paired_all_with_desc = paired_all_with_desc1 + paired_all_with_desc2
paired_all_with_desc = [data for data in paired_all_with_desc if data["label"] in ["CN", "AD"]]
```

```python
from monai.data import CacheDataset, DataLoader
from monai.data import CacheNTransDataset
import monai.data as md
from monai.data import CacheDataset, DataLoader
from monai.data import CacheNTransDataset
import monai.data as md
train_transforms=mt.Compose([
    mt.Lambdad(keys=["pet","mri"],func=mat_load),
    mt.EnsureChannelFirstd(keys=["pet","mri"],channel_dim='no_channel'),
    mt.Orientationd(keys=["pet","mri"],axcodes="LPI"),
    mt.CropForegroundd(keys=["pet","mri"],source_key="mri"),
    
    mt.HistogramNormalized(keys=["pet","mri"]),
    mt.ResizeWithPadOrCropd(keys=["pet","mri"],spatial_size=(160,192,160)),
    mt.Spacingd(keys=["pet","mri"],pixdim=(1.0,1.0,1.0), mode=("bilinear")),
    mt.NormalizeIntensityd(keys=["pet","mri"]),
    mt.ScaleIntensityd(keys=["pet","mri"],minv=0,maxv=1),
    mt.Identityd(keys=["label"]),
])
val_transforms=mt.Compose([
    mt.Lambdad(keys=["pet","mri"],func=mat_load),
    mt.EnsureChannelFirstd(keys=["pet","mri"],channel_dim='no_channel'),
    mt.Orientationd(keys=["pet","mri"],axcodes="LPI"),
    mt.CropForegroundd(keys=["pet","mri"],source_key="mri"),
    mt.HistogramNormalized(keys=["pet","mri"]),
    mt.ResizeWithPadOrCropd(keys=["pet","mri"],spatial_size=(160,192,160)),
    mt.Spacingd(keys=["pet","mri"],pixdim=(1.0,1.0,1.0), mode=("bilinear")),
    mt.NormalizeIntensityd(keys=["pet","mri"]),
    mt.ScaleIntensityd(keys=["pet","mri"],minv=0,maxv=1),
    mt.Identityd(keys=["label"]),
])
train_ds = CacheDataset(data=train_data, transform=train_transforms,num_workers=20)
train_loader = DataLoader(train_ds, batch_size=5, shuffle=True, num_workers=0)    

val_ds = CacheDataset(data=val_data, transform=val_transforms,num_workers=20)
val_loader = DataLoader(val_ds,  batch_size=5,shuffle=False, num_workers=0)
```

```python
import matplotlib.pyplot as plt
import torch
model = monai.networks.nets.DenseNet121(spatial_dims=3, in_channels=1, out_channels=2).to(device)

loss_function = torch.nn.CrossEntropyLoss()  # CrossEntropy expects class indices, not one-hot encoded labels
optimizer = torch.optim.Adam(model.parameters(), 1e-5)

# start a typical PyTorch training
val_interval = 2
best_metric = -1
best_metric_epoch = -1
epoch_loss_values = []
metric_values = []
writer = SummaryWriter()
max_epochs =200

for epoch in range(max_epochs):
    print("-" * 10)
    print(f"epoch {epoch + 1}/{max_epochs}")
    model.train()
    epoch_loss = 0
    step = 0

    for batch_data in train_loader:
        step += 1
        # Extract MRI data and labels
        mri_images = batch_data["mri"].to(device)  # Use only MRI images as input
        labels = torch.tensor(batch_data["label"], dtype=torch.long).to(device)  # Convert labels to Long type

        # Convert one-hot encoded labels to class indices if necessary
        if labels.dim() == 2:  # If the labels are 2D (one-hot), convert to class indices
            labels = torch.argmax(labels, dim=1)

        optimizer.zero_grad()
        outputs = model(mri_images)
        loss = loss_function(outputs, labels)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()
        epoch_len = len(train_ds) // train_loader.batch_size
        print(f"{step}/{epoch_len}, train_loss: {loss.item():.4f}")
        writer.add_scalar("train_loss", loss.item(), epoch_len * epoch + step)

    epoch_loss /= step
    epoch_loss_values.append(epoch_loss)
    print(f"epoch {epoch + 1} average loss: {epoch_loss:.4f}")

    # 验证部分每隔val_interval进行一次
    if (epoch + 1) % val_interval == 0:
        model.eval()
        num_correct = 0.0
        metric_count = 0
        for val_data in val_loader:
            val_images = val_data["mri"].to(device)
            val_labels = torch.tensor(val_data["label"], dtype=torch.long).to(device)

            if val_labels.dim() == 2:
                val_labels = torch.argmax(val_labels, dim=1)

            with torch.no_grad():
                val_outputs = model(val_images)
                value = torch.eq(val_outputs.argmax(dim=1), val_labels)
                metric_count += len(value)
                num_correct += value.sum().item()

        # 计算并保存准确率
        metric = num_correct / metric_count
        metric_values.append(metric)

        if metric > best_metric:
            best_metric = metric
            best_metric_epoch = epoch + 1
            torch.save(model.state_dict(), "best_metric_model_classification3d_array.pth")
            print("saved new best metric model")

        print(f"Current epoch: {epoch+1} current accuracy: {metric:.4f} ")
        print(f"Best accuracy: {best_metric:.4f} at epoch {best_metric_epoch}")
        writer.add_scalar("val_accuracy", metric, epoch + 1)

# 训练完成后的打印信息
print(f"Training completed, best_metric: {best_metric:.4f} at epoch: {best_metric_epoch}")
writer.close()

# 打印收集的训练损失和验证准确率
print("Training Loss Values: ", epoch_loss_values)
print("Validation Accuracy Values: ", metric_values)

# 生成曲线图
epochs = range(1, len(epoch_loss_values) + 1)
val_epochs = range(val_interval, max_epochs + 1, val_interval)

# 绘制图像
plt.figure(figsize=(12, 5))

# 绘制训练损失曲线
plt.subplot(1, 2, 1)
plt.plot(epochs, epoch_loss_values, label='Training Loss', marker='o')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('Training Loss over Epochs')
plt.legend()

# 绘制验证准确率曲线
plt.subplot(1, 2, 2)
plt.plot(val_epochs, metric_values, label='Validation Accuracy', marker='o')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.title('Validation Accuracy over Epochs')
plt.legend()

# 调整布局
plt.tight_layout()
plt.show()


```

# mri判断AD CN

```python
from sklearn.metrics import confusion_matrix, f1_score, matthews_corrcoef, roc_auc_score
import matplotlib.pyplot as plt
import torch
# Create DenseNet121, CrossEntropyLoss and Adam optimizer
model = monai.networks.nets.DenseNet121(spatial_dims=3, in_channels=1, out_channels=2).to(device)

loss_function = torch.nn.CrossEntropyLoss()  # CrossEntropy expects class indices, not one-hot encoded labels
optimizer = torch.optim.Adam(model.parameters(), 1e-4)

# start a typical PyTorch training
val_interval =2
best_metric = -1
best_metric_epoch = -1
writer = SummaryWriter()
max_epochs =200


# 初始化存储训练损失和验证指标的列表
epoch_loss_values = []
accuracy_values = []
sensitivity_values = []
specificity_values = []
f1_score_values = []
mcc_values = []
auc_values = []

for epoch in range(max_epochs):
    print("-" * 10)
    print(f"epoch {epoch + 1}/{max_epochs}")
    model.train()
    epoch_loss = 0
    step = 0

    for batch_data in train_loader:
        step += 1
        mri_images = batch_data["mri"].to(device)
        labels = torch.tensor(batch_data["label"], dtype=torch.long).to(device)

        if labels.dim() == 2:  # 转换one-hot编码标签为类索引
            labels = torch.argmax(labels, dim=1)

        optimizer.zero_grad()
        outputs = model(mri_images)
        loss = loss_function(outputs, labels)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()

    # 计算并保存平均训练损失
    epoch_loss /= step
    epoch_loss_values.append(epoch_loss)
    print(f"epoch {epoch + 1} average loss: {epoch_loss:.4f}")

    # 验证部分每隔val_interval进行一次
    if (epoch + 1) % val_interval == 0:
        model.eval()
        y_true = []
        y_pred = []
        y_prob = []
        
        for val_data in val_loader:
            val_images = val_data["mri"].to(device)
            val_labels = torch.tensor(val_data["label"], dtype=torch.long).to(device)

            if val_labels.dim() == 2:
                val_labels = torch.argmax(val_labels, dim=1)

            with torch.no_grad():
                val_outputs = model(val_images)
                preds = val_outputs.argmax(dim=1)
                y_true.extend(val_labels.cpu().numpy())
                y_pred.extend(preds.cpu().numpy())
                y_prob.extend(val_outputs[:, 1].cpu().numpy())  # 概率用于 AUC 计算

        # 计算混淆矩阵
        cm = confusion_matrix(y_true, y_pred)
        TN, FP, FN, TP = cm.ravel()  # 从混淆矩阵中提取 TN, FP, FN, TP

        # 计算各项指标
        acc = (TP + TN) / (TP + TN + FP + FN)  # 准确率
        sen = TP / (TP + FN)  # 召回率/敏感性
        spe = TN / (TN + FP)  # 特异性
        f1 = f1_score(y_true, y_pred)  # F1分数
        mcc = matthews_corrcoef(y_true, y_pred)  # 马修斯相关系数
        auc = roc_auc_score(y_true, y_prob)  # AUC

        # 保存每个指标的值
        accuracy_values.append(acc)
        sensitivity_values.append(sen)
        specificity_values.append(spe)
        f1_score_values.append(f1)
        mcc_values.append(mcc)
        auc_values.append(auc)

        print(f"ACC: {acc:.4f}, SEN: {sen:.4f}, SPE: {spe:.4f}, F1: {f1:.4f}, MCC: {mcc:.4f}, AUC: {auc:.4f}")

        # 保存模型
        if acc > best_metric:
            best_metric = acc
            best_metric_epoch = epoch + 1
            torch.save(model.state_dict(), "best_metric_model_classification3d_array.pth")
            print("saved new best metric model")

# 打印训练完成的结果
print(f"Training completed, best_metric: {best_metric:.4f} at epoch: {best_metric_epoch}")
writer.close()

# 打印收集的指标
print("Training Loss Values: ", epoch_loss_values)
print("Validation Accuracy Values: ", accuracy_values)
print("Validation Sensitivity (Recall) Values: ", sensitivity_values)
print("Validation Specificity Values: ", specificity_values)
print("F1 Score Values: ", f1_score_values)
print("MCC Values: ", mcc_values)
print("AUC Values: ", auc_values)

# 生成图像显示训练损失和各指标变化
epochs = range(1, len(epoch_loss_values) + 1)
val_epochs = range(val_interval, max_epochs + 1, val_interval)

# 绘制图像
plt.figure(figsize=(12, 8))

# 绘制训练损失曲线
plt.subplot(2, 2, 1)
plt.plot(epochs, epoch_loss_values, label='Training Loss', marker='o')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('Training Loss over Epochs')
plt.legend()

# 绘制验证准确率曲线
plt.subplot(2, 2, 2)
plt.plot(val_epochs, accuracy_values, label='Validation Accuracy', marker='o')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.title('Validation Accuracy over Epochs')
plt.legend()

# 绘制 F1 Score 曲线
plt.subplot(2, 2, 3)
plt.plot(val_epochs, f1_score_values, label='F1 Score', marker='o')
plt.xlabel('Epochs')
plt.ylabel('F1 Score')
plt.title('F1 Score over Epochs')
plt.legend()

# 绘制 AUC 曲线
plt.subplot(2, 2, 4)
plt.plot(val_epochs, auc_values, label='AUC', marker='o')
plt.xlabel('Epochs')
plt.ylabel('AUC')
plt.title('AUC over Epochs')
plt.legend()

# 调整布局
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, f1_score, matthews_corrcoef, roc_auc_score
import matplotlib.pyplot as plt
import torch
# Create DenseNet121, CrossEntropyLoss and Adam optimizer
model = monai.networks.nets.DenseNet121(spatial_dims=3, in_channels=2, out_channels=2).to(device)

loss_function = torch.nn.CrossEntropyLoss()  # CrossEntropy expects class indices, not one-hot encoded labels
optimizer = torch.optim.Adam(model.parameters(), 1e-4)

# start a typical PyTorch training
val_interval =2
best_metric = -1
best_metric_epoch = -1
writer = SummaryWriter()
max_epochs =200


# 初始化存储训练损失和验证指标的列表
epoch_loss_values = []
accuracy_values = []
sensitivity_values = []
specificity_values = []
f1_score_values = []
mcc_values = []
auc_values = []

for epoch in range(max_epochs):
    print("-" * 10)
    print(f"epoch {epoch + 1}/{max_epochs}")
    model.train()
    epoch_loss = 0
    step = 0

    for batch_data in train_loader:
        step += 1
        mri_images = batch_data["mri"].to(device)
        pet_images = batch_data["pet"].to(device)
        labels = torch.tensor(batch_data["label"], dtype=torch.long).to(device)
        combined = torch.cat((mri_images, pet_images ), dim=1)
        

        if labels.dim() == 2:  # 转换one-hot编码标签为类索引
            labels = torch.argmax(labels, dim=1)

        optimizer.zero_grad()
        outputs = model(combined)
        loss = loss_function(outputs, labels)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()

    # 计算并保存平均训练损失
    epoch_loss /= step
    epoch_loss_values.append(epoch_loss)
    print(f"epoch {epoch + 1} average loss: {epoch_loss:.4f}")

    # 验证部分每隔val_interval进行一次
    if (epoch + 1) % val_interval == 0:
        model.eval()
        y_true = []
        y_pred = []
        y_prob = []
        
        for val_data in val_loader:
            val_mri_images = val_data["mri"].to(device)
            val_pet_images = val_data["pet"].to(device)
            val_labels = torch.tensor(val_data["label"], dtype=torch.long).to(device)
            val_combined = torch.cat((val_mri_images, val_pet_images), dim=1)
            if val_labels.dim() == 2:
                val_labels = torch.argmax(val_labels, dim=1)

            with torch.no_grad():
                val_outputs = model(val_combined)
                preds = val_outputs.argmax(dim=1)
                y_true.extend(val_labels.cpu().numpy())
                y_pred.extend(preds.cpu().numpy())
                y_prob.extend(val_outputs[:, 1].cpu().numpy())  # 概率用于 AUC 计算

        # 计算混淆矩阵
        cm = confusion_matrix(y_true, y_pred)
        TN, FP, FN, TP = cm.ravel()  # 从混淆矩阵中提取 TN, FP, FN, TP

        # 计算各项指标
        acc = (TP + TN) / (TP + TN + FP + FN)  # 准确率
        sen = TP / (TP + FN)  # 召回率/敏感性
        spe = TN / (TN + FP)  # 特异性
        f1 = f1_score(y_true, y_pred)  # F1分数
        mcc = matthews_corrcoef(y_true, y_pred)  # 马修斯相关系数
        auc = roc_auc_score(y_true, y_prob)  # AUC

        # 保存每个指标的值
        accuracy_values.append(acc)
        sensitivity_values.append(sen)
        specificity_values.append(spe)
        f1_score_values.append(f1)
        mcc_values.append(mcc)
        auc_values.append(auc)

        print(f"ACC: {acc:.4f}, SEN: {sen:.4f}, SPE: {spe:.4f}, F1: {f1:.4f}, MCC: {mcc:.4f}, AUC: {auc:.4f}")

        # 保存模型
        if acc > best_metric:
            best_metric = acc
            best_metric_epoch = epoch + 1
            torch.save(model.state_dict(), "best_metric_model_classification3d_array.pth")
            print("saved new best metric model")

# 打印训练完成的结果
print(f"Training completed, best_metric: {best_metric:.4f} at epoch: {best_metric_epoch}")
writer.close()

# 打印收集的指标
print("Training Loss Values: ", epoch_loss_values)
print("Validation Accuracy Values: ", accuracy_values)
print("Validation Sensitivity (Recall) Values: ", sensitivity_values)
print("Validation Specificity Values: ", specificity_values)
print("F1 Score Values: ", f1_score_values)
print("MCC Values: ", mcc_values)
print("AUC Values: ", auc_values)

# 生成图像显示训练损失和各指标变化
epochs = range(1, len(epoch_loss_values) + 1)
val_epochs = range(val_interval, max_epochs + 1, val_interval)

# 绘制图像
plt.figure(figsize=(12, 8))

# 绘制训练损失曲线
plt.subplot(2, 2, 1)
plt.plot(epochs, epoch_loss_values, label='Training Loss', marker='o')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('Training Loss over Epochs')
plt.legend()

# 绘制验证准确率曲线
plt.subplot(2, 2, 2)
plt.plot(val_epochs, accuracy_values, label='Validation Accuracy', marker='o')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.title('Validation Accuracy over Epochs')
plt.legend()

# 绘制 F1 Score 曲线
plt.subplot(2, 2, 3)
plt.plot(val_epochs, f1_score_values, label='F1 Score', marker='o')
plt.xlabel('Epochs')
plt.ylabel('F1 Score')
plt.title('F1 Score over Epochs')
plt.legend()

# 绘制 AUC 曲线
plt.subplot(2, 2, 4)
plt.plot(val_epochs, auc_values, label='AUC', marker='o')
plt.xlabel('Epochs')
plt.ylabel('AUC')
plt.title('AUC over Epochs')
plt.legend()

# 调整布局
plt.tight_layout()
plt.show()
```
